```python
from Bio import SeqIO
from Bio.Seq import Seq

# Load the FASTA file
records = list(SeqIO.parse(r"C:\Users\Akintayo\Downloads\Bioinfornatics\dna2.fasta", "fasta"))

print("Number of sequences:", len(records))

for i, record in enumerate(records, start=0):
    print(f"\nSequence {i}")
    print("ID:", record.id)
    print("Description:", record.description)
    print("Length:", len(record.seq))
```

```python
def find_orfs(sequence, frame):         ## Create a function that receives a DNA sequence and a reading frame.
    
    orfs = []
    
    start_codon = "ATG"
    stop_codons = {"TAA", "TAG", "TGA"}
    
    start = frame - 1           ## Adjust the frame to be zero-based for indexing.
    
    i = start
    
    while i <= len(sequence) - 3:
        
        codon = sequence[i:i+3]
        
        if codon == start_codon:
            
            orf_start = i
            
            j = i + 3
            
            while j <= len(sequence) - 3:
                
                stop = sequence[j:j+3]
                
                if stop in stop_codons:
                    
                    orf_end = j + 3
                    
                    length = orf_end - orf_start
                    
                    orfs.append({
                        "start": orf_start + 1,
                        "end": orf_end,
                        "length": length,
                        "sequence": sequence[orf_start:orf_end]
                    })
                    
                    break
                
                j += 3
        
        i += 3
    
    return orfs
```

## What is the length of the longest ORF appearing in reading frame 2 of any of the sequences?

```python
longest_orf = None

for record in records:
    
    sequence = str(record.seq)
    
    orfs = find_orfs(sequence, 2)
    
    for orf in orfs:
        
        if longest_orf is None or orf["length"] > longest_orf["length"]:
            
            longest_orf = {
                "sequence_id": record.id,
                **orf
            }

print(longest_orf)
```

## What is the length of the longest ORF appearing in any sequence and in any forward reading frame?

```python
longest_forward_orf = None

for record in records:
    
    sequence = str(record.seq)
    
    for frame in [1, 2, 3]:
        
        orfs = find_orfs(sequence, frame)
        
        for orf in orfs:
            
            if (longest_forward_orf is None or
                orf["length"] > longest_forward_orf["length"]):
                
                longest_forward_orf = {
                    "sequence_id": record.id,
                    "frame": frame,
                    **orf
                }

print(longest_forward_orf)
```

```python
## COUNTING 6-BASE REPEATS
sequence = "ATGCGCGCGCTA"

k = 6

repeats = [
    sequence[i:i+k]
    for i in range(len(sequence) - k + 1)
]

print(repeats)
```

```python
print(counts.most_common(1))
```

```python
## in all sequences
from collections import Counter

k = 6

all_repeats = []

for record in records:
    
    sequence = str(record.seq)
    
    for i in range(len(sequence) - k + 1):
        
        repeat = sequence[i:i+k]
        
        all_repeats.append(repeat)

counts = Counter(all_repeats)

most_common_repeat, frequency = counts.most_common(1)[0]

print("Most frequent repeat:", most_common_repeat)
print("Number of occurrences:", frequency)
```

## Find all repeats of length 12 in the input file.

```python
k = 12

all_repeats = []

for record in records:
    
    sequence = str(record.seq)
    
    for i in range(len(sequence) - k + 1):
        
        repeat = sequence[i:i+k]
        
        all_repeats.append(repeat)

counts = Counter(all_repeats)
```

```python
max_count = max(counts.values())

print("Maximum number of occurrences:", max_count)
```

```python
max_repeats = [
    repeat
    for repeat, count in counts.items()
    if count == max_count
]

print("Max =", max_count)

print("Number of different repeats occurring Max times:",
      len(max_repeats))

for repeat in max_repeats:
    print(repeat)
```
